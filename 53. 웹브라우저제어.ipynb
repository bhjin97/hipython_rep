# 웹브라우저 자동화

```python
import selenium
print(selenium.__version__)

#Selenium의 웹드라이버 기능 불러오기 (크롬, 파이어폭스 등 조작 가능)
from selenium import webdriver

#크롬 브라우저 설정 옵션을 사용하기 위한 클래스
from selenium.webdriver.chrome.options import Options


#크롬 실행 옵션 객체 생성
chrome_options = Options()

# 브라우저 종료 방지 옵션
chrome_options.add_experimental_option("detach", True)

#설정된 옵션과 함께 크롬 브라우저를 실행
driver = webdriver.Chrome(options=chrome_options)

#지정한 URL을 자동으로 열기 
driver.get('https://www.python.org')
```

```python
search_form = driver.find_element(by='id', value='id-search-field')
print(search_form)
```

```python
from selenium.webdriver import Chrome
import os

file_path = os.getcwd() + '/html/basic2.html'
url = 'file:///'+file_path

driver = webdriver.Chrome()
driver.get(url)
```

```python
username_el = driver.find_element(by='name', value='username')
username_el.send_keys('bae')
```

```python
pass_el = driver.find_element(by='name', value='password')
pass_el.send_keys('abc123')
```

```python
# 파이썬 전문가그룹의 링크값
link = driver.find_elements(by = 'link text', value ='파이썬전문가그룹')
link[0].click() # 링크로 이동함
```

```python
driver.back() # 원래 웹으로 돌아옴
```
