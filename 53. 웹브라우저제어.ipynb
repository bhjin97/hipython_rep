# 웹브라우저 자동화

```python
import selenium
print(selenium.__version__)

#Selenium의 웹드라이버 기능 불러오기 (크롬, 파이어폭스 등 조작 가능)
from selenium import webdriver

#크롬 브라우저 설정 옵션을 사용하기 위한 클래스
from selenium.webdriver.chrome.options import Options


#크롬 실행 옵션 객체 생성
chrome_options = Options()

# 브라우저 종료 방지 옵션
chrome_options.add_experimental_option("detach", True)

#설정된 옵션과 함께 크롬 브라우저를 실행
driver = webdriver.Chrome(options=chrome_options)

#지정한 URL을 자동으로 열기 
driver.get('https://www.python.org')
```

```python
search_form = driver.find_element(by='id', value='id-search-field')
print(search_form)
```

# 검색 결과 가져오기

```python
from selenium import webdriver

driver = webdriver.Chrome()
driver.get('https://www.python.org')
```

```python
from selenium.webdriver.common.by import By
input_el = driver.find_element(by = By.ID, value ='id-search-field')
input_el.send_keys('pycon')
```

```python
from selenium.webdriver.common.keys import Keys
input_el.send_keys(Keys.RETURN)
```

```python
link = driver.find_element(By.XPATH, '/html/body/div/div[3]/div/section/form/ul/li[2]/h3/a')
link.click()
```

```python
link = driver.find_elements(by = 'css selector', value = 'form li h3 >a')
for i in link:
    print(i.text)
```

```python
from selenium import webdriver

driver = webdriver.Chrome()
driver.get('https://www.python.org')
from selenium.webdriver.common.by import By
input_el = driver.find_element(by = By.ID, value ='id-search-field')
input_el.send_keys('pycon')
from selenium.webdriver.common.keys import Keys
input_el.send_keys(Keys.RETURN)

# 암묵적 대기 설정,페이지가 로드될 때까지 최대 3초 대기
driver.implicitly_wait(3)  
link = driver.find_elements(by = 'css selector', value = 'form li h3 >a')
for i in link:
    print(i.text)
```
