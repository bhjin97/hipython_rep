```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "year": [2010, 2016, 2018, 2020],      
    "sales": [9, 30, 40, 45]                  
}
df = pd.DataFrame(data)

# ── 2) 기본 스타일 설정 ──
sns.set(style="whitegrid", font="Malgun Gothic")  # 한글 폰트: Windows=Malgun Gothic, macOS=AppleGothic 등

# ── 3) 그래프 그리기 ──
plt.figure(figsize=(8,5))
sns.barplot(
    data=df,
    x="year", y="sales",
    color="#62deac"       
)

# 값 라벨 표시
for i, v in enumerate(df["sales"]):
    plt.text(i, v, f"{v}", ha="center", va="bottom", fontsize=13, fontweight='bold')

plt.title("연도별 매출액 추이 (단위: 조 원)", fontsize=20, fontweight='bold')
plt.xlabel("년도", fontsize=13, fontweight='bold')
plt.ylabel("매출액", fontsize=13, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ── 1) 데이터 입력 ──
data = {
    "year": [2016, 2017, 2018, 2019],
    "sales": [5, 8, 9, 12]   # 조 단위 값
}
df = pd.DataFrame(data)

# ── 2) 스타일 설정 ──
sns.set(style="whitegrid", font="Malgun Gothic")  # macOS는 AppleGothic

# ── 3) 가로 막대 그래프 ──
plt.figure(figsize=(8,5))
sns.barplot(
    data=df,
    x="sales", y="year",
    color="#43d69c" ,
    orient='h'           # ← 가로 막대를 확실히 지정
)

# 값 라벨 표시 (막대 끝 오른쪽에)
for i, v in enumerate(df["sales"]):
    plt.text(v, i, f"{v}조", ha="left", va="center", fontsize=13, fontweight='bold')

plt.title("화장품 온라인 쇼핑 거래액 (단위: 조 원)", fontsize=20, fontweight='bold')
plt.xlabel("거래액 (조 원)", fontsize=13, fontweight='bold')
plt.ylabel("년도", fontsize=13, fontweight='bold')

plt.tight_layout()
plt.show()

```
